```python
def add(a, b):
    return a + b
```

```python
def sum_list(values):
    total = 0
    for v in range(len(values)):
        total = add(values[v], total)
    return total
```

```python
values = [1,2,4,5]
sum_list(values)
```

```python
values = [3,6]
sum_list(values)
```

```python
values = [4,6,2,8,0,-1]
sum_list(values)
```
