```python
def subtract(a,b):

    return a - b

def distance_from_zero(a):
    b = 0
    x = subtract(a, b)
    distance = abs(x)
    return distance
```

```python
print(subtract(1, 5))
print(subtract(7, 2))
print(subtract(-2, 3))

print(distance_from_zero(2))
print(distance_from_zero(5))
print(distance_from_zero(-6))
```
